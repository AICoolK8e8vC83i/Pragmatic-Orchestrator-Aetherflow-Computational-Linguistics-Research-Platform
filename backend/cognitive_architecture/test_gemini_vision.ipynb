# Gemini 2.5 Flash Vision Test

Testing Gemini 2.5 Flash with local Chinese calligraphy image.


```python
%pip install google-generativeai python-dotenv pillow

import google.generativeai as genai
import os
from dotenv import load_dotenv
from PIL import Image

load_dotenv()
```

```python
# API Configuration
api_key = os.getenv("GEMINI_API_KEY") or os.getenv("GOOGLE_AI_API_KEY")
model_name = "gemini-2.5-flash"

if not api_key:
    print("❌ GEMINI_API_KEY not found in environment")
    print("\nPlease set the API key:")
    print("export GEMINI_API_KEY=your_key_here")
    print("\nOr add to .env file:")
    print("GEMINI_API_KEY=your_key_here")
else:
    print(f"✅ API key found")
    print(f"Testing model: {model_name}")
```

```python
# Initialize model using the simpler API (genai.GenerativeModel)
if api_key:
    try:
        genai.configure(api_key=api_key)
        model = genai.GenerativeModel(model_name)
        print("✅ Model initialized successfully")
    except Exception as e:
        print(f"❌ Model initialization failed: {e}")
        model = None
        import traceback
        traceback.print_exc()
else:
    model = None
```

```python
# Test with local Chinese calligraphy image
image_path = "Chinese cailgraphy.jpg"
prompt = "Analyze this image and describe what you see in detail."

print("Testing Gemini 2.5 Flash with local image...")
print(f"Image path: {image_path}")
print(f"Prompt: {prompt}")
print("\n" + "="*70)

if model:
    try:
        # Check if file exists
        if not os.path.exists(image_path):
            print(f"❌ Image file not found: {image_path}")
            print(f"Current directory: {os.getcwd()}")
        else:
            print(f"✅ Image file found: {image_path}")
            
            # Load local image using PIL
            img = Image.open(image_path)
            print(f"✅ Image loaded: {img.size[0]}x{img.size[1]} pixels, mode: {img.mode}")
            
            # Generate content with image and text
            print("\n🔄 Calling Gemini API...")
            response = model.generate_content(
                [prompt, img],
                generation_config=genai.types.GenerationConfig(
                    temperature=0.7,
                    max_output_tokens=2000,
                )
            )
            
            print("\n✅ Response:")
            print(response.text)
            
    except Exception as e:
        print(f"❌ Error: {e}")
        import traceback
        traceback.print_exc()
else:
    print("⚠️  Model not initialized. Check API key.")
```

```python
# Test with the newer genai.Client API approach
from google.genai import types

print("\n" + "="*70)
print("Testing with newer genai.Client API...")
print("="*70)

if api_key:
    try:
        # Initialize client
        client = genai.Client(api_key=api_key)
        model_obj = client.models.get(model_name)
        print(f"✅ Client initialized, model: {model_name}")
        
        # Load image
        image_path = "Chinese cailgraphy.jpg"
        if os.path.exists(image_path):
            with open(image_path, 'rb') as f:
                image_data = f.read()
            
            # Create Part from bytes
            image_part = types.Part.from_bytes(
                data=image_data,
                mime_type="image/jpeg"
            )
            
            prompt = "Analyze this image and describe what you see in detail."
            
            print("\n🔄 Calling Gemini API (newer API)...")
            response = model_obj.generate_content(
                contents=[image_part, prompt],
                config=types.GenerateContentConfig(
                    temperature=0.7,
                    max_output_tokens=2000,
                )
            )
            
            print("\n✅ Response:")
            print(response.text)
        else:
            print(f"❌ Image file not found: {image_path}")
            
    except Exception as e:
        print(f"❌ Error with newer API: {e}")
        import traceback
        traceback.print_exc()
else:
    print("⚠️  API key not found")
```

```python
# Test with cultural preservation prompt (as used in code_switcher)
image_path = "Chinese cailgraphy.jpg"
cultural_prompt = """Deconstruct the 语用学 (pragmatics) of this term's visual representation—why does the character convey **unspoken social contract**? 
Analyze the calligraphy style, cultural significance, and any text visible in the image."""

print("\n" + "="*70)
print("Testing with cultural preservation prompt...")
print("="*70)

if model:
    try:
        if os.path.exists(image_path):
            img = Image.open(image_path)
            
            print("\n🔄 Calling Gemini API with cultural prompt...")
            response = model.generate_content(
                [cultural_prompt, img],
                generation_config=genai.types.GenerationConfig(
                    temperature=0.7,
                    max_output_tokens=2000,
                )
            )
            
            print("\n✅ Response:")
            print(response.text)
        else:
            print(f"❌ Image file not found: {image_path}")
            
    except Exception as e:
        print(f"❌ Error: {e}")
        import traceback
        traceback.print_exc()
else:
    print("⚠️  Model not initialized")
```
