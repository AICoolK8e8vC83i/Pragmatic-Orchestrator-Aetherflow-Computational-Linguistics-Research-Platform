Testing Qwen3-VL 2B via Ollama for vision/document understanding.

```python
import ollama
import base64
import os
from dotenv import load_dotenv
from pathlib import Path

load_dotenv()
```

```python
# Model configuration
model_name = "qwen3-vl:2b"

print(f"Testing model: {model_name}")
```

```python
# Helper function to encode image to base64
def encode_image(image_path):
    """Encode image file to base64 string."""
    with open(image_path, "rb") as image_file:
        return base64.b64encode(image_file.read()).decode('utf-8')
```

## Upload Your Image

1. Upload an image file to this directory or provide the path
2. The model will analyze it and describe what's happening


```python
# Set your image path here
# Example: image_path = "path/to/your/image.jpg"
# Or use: image_path = "test_image.png"

image_path = "/Users/kc/Project-AETHERFLOW-Multi-AI-Agent-Productivity-Intelligent-Research-System/backend/cognitive_architecture/Chinese cailgraphy.jpg"

# Check if file exists
if os.path.exists(image_path):
    print(f"✅ Image found: {image_path}")
    image_base64 = encode_image(image_path)
    print(f"✅ Image encoded (length: {len(image_base64)} chars)")
else:
    print(f"❌ Image not found: {image_path}")
    print("\nPlease:")
    print("1. Upload an image to this directory")
    print("2. Update the image_path variable above")
    image_base64 = None
```

```python
# Vision test with image
if image_base64:
    prompt = "Describe what's happening in this image in detail."
    
    print(f"Prompt: {prompt}")
    print("\n" + "="*70)
    
    try:
        # Qwen3-VL uses images via ollama's vision API
        # Note: You may need to use ollama.chat() with images parameter
        # For now, testing basic functionality
        response = ollama.generate(
            model=model_name,
            prompt=prompt,
            images=[image_base64]
        )
        
        print("✅ Response:")
        print(response.get("response", "No response"))
        
    except Exception as e:
        print(f"❌ Error: {e}")
        print("\nNote: Qwen3-VL may need specific image format.")
        print("Try using ollama.chat() with images parameter for vision tasks.")
else:
    print("⚠️  No image provided. Please upload an image first.")
```

```python
# Alternative: Test with text-only first
text_prompt = "Say 'Qwen3-VL 2B is working!' and explain your vision capabilities."

print("Testing text-only (fallback)...")
print("\n" + "="*70)

try:
    response = ollama.generate(
        model=model_name,
        prompt=text_prompt,
        options={
            "temperature": 0.7,
            "top_p": 0.9
        }
    )
    
    print("✅ Response:")
    print(response.get("response", "No response"))
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nMake sure:")
    print("1. Ollama is running: ollama serve")
    print(f"2. Model is pulled: ollama pull {model_name}")
```
