# DeepSeek R1 1.5B Test

Testing DeepSeek R1 1.5B via Ollama for meta-routing and reasoning.


```python
import ollama
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Model configuration
model_name = "deepseek-r1:1.5b"

print(f"Testing model: {model_name}")
```

```python
# Simple test prompt
prompt = "Say 'DeepSeek R1 1.5B is working!' and explain what you're good at."

print(f"Prompt: {prompt}")
print("\n" + "="*70)
```

```python
# Generate response
try:
    response = ollama.generate(
        model=model_name,
        prompt=prompt,
        options={
            "temperature": 0.7,
            "top_p": 0.9
        }
    )
    
    print("✅ Response:")
    print(response.get("response", "No response"))
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nMake sure:")
    print("1. Ollama is running: ollama serve")
    print(f"2. Model is pulled: ollama pull {model_name}")
```

```python
# Meta-routing test (what DeepSeek is designed for)
routing_prompt = """You are a meta-router for an AI agent system. 
Given this task: 'Research the latest developments in multi-agent AI systems'

Which model should be called first?
Options: claude_4_5_sonnet, gpt_5_mini, sonar, or claude_4_5_haiku

Respond in JSON format:
{
    "model": "model_name",
    "reasoning": "Why this model"
}"""

print("Testing meta-routing capability...")
print("\n" + "="*70)

try:
    response = ollama.generate(
        model=model_name,
        prompt=routing_prompt,
        options={
            "temperature": 0.3,  # Lower temp for routing decisions
            "top_p": 0.9
        }
    )
    
    print("✅ Routing Decision:")
    print(response.get("response", "No response"))
    
except Exception as e:
    print(f"❌ Error: {e}")
```
