# Claude Sonnet 4.5 Test

Testing Claude 4.5 Sonnet (SOTA for general/coding tasks) via Anthropic API.


```python
import anthropic
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# API configuration
api_key = os.getenv("ANTHROPIC_API_KEY")
model_name = "claude-sonnet-4-5-20250929"

if not api_key:
    print("❌ ANTHROPIC_API_KEY not found in .env file")
    print("\nPlease add to .env:")
    print("ANTHROPIC_API_KEY=sk-ant-...")
else:
    print(f"✅ API key found")
    print(f"Testing model: {model_name}")
```

```python
# Initialize client
if api_key:
    client = anthropic.Anthropic(api_key=api_key)
    print("✅ Client initialized")
else:
    client = None
```

```python
# Simple test prompt
prompt = "Say 'Claude 4.5 Sonnet is working!' and explain what you're best at."

print(f"Prompt: {prompt}")
print("\n" + "="*70)
```

```python
# Generate response
if client:
    try:
        message = client.messages.create(
            model=model_name,
            max_tokens=300,
            temperature=0.7,
            messages=[
                {"role": "user", "content": prompt}
            ]
        )
        
        response_text = message.content[0].text
        print("✅ Response:")
        print(response_text)
        
    except Exception as e:
        print(f"❌ Error: {e}")
        print("\nCheck:")
        print("1. API key is valid")
        print("2. Model name is correct: claude-4.5-sonnet")
        print("3. You have API credits")
else:
    print("⚠️  Client not initialized. Check API key.")
```

```python
# Coding task test (what Sonnet 4.5 excels at)
coding_prompt = """Write a Python function that takes a list of numbers and returns 
the sum of all even numbers. Include a docstring and example usage."""

print("Testing coding capabilities...")
print("\n" + "="*70)

if client:
    try:
        message = client.messages.create(
            model=model_name,
            max_tokens=500,
            temperature=0.2,  # Lower temp for code
            messages=[
                {"role": "user", "content": coding_prompt}
            ]
        )
        
        response_text = message.content[0].text
        print("✅ Response:")
        print(response_text)
        
    except Exception as e:
        print(f"❌ Error: {e}")
```
