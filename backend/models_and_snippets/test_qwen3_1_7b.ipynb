# Qwen 3 1.7B Test

Testing Qwen 3 1.7B via Ollama for general tasks and reasoning.


```python
import ollama
import os
from dotenv import load_dotenv

load_dotenv()
```

```python
# Model configuration
model_name = "qwen3:1.7b"

print(f"Testing model: {model_name}")
```

```python
# Simple test prompt
prompt = "Say 'Qwen 3 1.7B is working!' and explain what you're good at."

print(f"Prompt: {prompt}")
print("\n" + "="*70)
```

```python
# Generate response
try:
    response = ollama.generate(
        model=model_name,
        prompt=prompt,
        options={
            "temperature": 0.7,
            "top_p": 0.9
        }
    )
    
    print("✅ Response:")
    print(response.get("response", "No response"))
    
except Exception as e:
    print(f"❌ Error: {e}")
    print("\nMake sure:")
    print("1. Ollama is running: ollama serve")
    print(f"2. Model is pulled: ollama pull {model_name}")
```

```python
# General reasoning test
reasoning_prompt = """Explain the difference between LangGraph, CrewAI, and AutoGen 
for multi-agent systems. Keep it concise (3-4 sentences)."""

print("Testing general reasoning...")
print("\n" + "="*70)

try:
    response = ollama.generate(
        model=model_name,
        prompt=reasoning_prompt,
        options={
            "temperature": 0.7,
            "top_p": 0.9
        }
    )
    
    print("✅ Response:")
    print(response.get("response", "No response"))
    
except Exception as e:
    print(f"❌ Error: {e}")
```
