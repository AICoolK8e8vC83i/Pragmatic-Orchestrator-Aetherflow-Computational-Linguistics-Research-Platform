```python
# Debug: Check audio file properties
try:
    import librosa
    
    print("Checking audio file properties...")
    print("="*70)
    
    # Load audio to check properties
    audio_data, sample_rate = librosa.load(speech_file, sr=48000)
    
    print(f"✅ File loaded successfully")
    print(f"Duration: {len(audio_data) / sample_rate:.2f} seconds")
    print(f"Sample rate: {sample_rate} Hz")
    print(f"Channels: {'Mono' if len(audio_data.shape) == 1 else 'Stereo'}")
    print(f"Audio data range: [{audio_data.min():.3f}, {audio_data.max():.3f}]")
    print(f"Non-zero samples: {(audio_data != 0).sum()} / {len(audio_data)}")
    
    if (audio_data == 0).all():
        print("\n⚠️  WARNING: Audio file appears to be silent!")
    elif (audio_data != 0).sum() < len(audio_data) * 0.01:
        print("\n⚠️  WARNING: Audio file appears to be mostly silent!")
    else:
        print("\n✅ Audio file appears to have content")
        
except ImportError:
    print("⚠️  librosa not installed. Install with: pip install librosa")
except Exception as e:
    print(f"❌ Error checking audio: {e}")
```
